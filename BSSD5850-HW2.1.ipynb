```python
import random 
import numpy as np
```

```python
def weighted_sum(vec, weights):
    w_vec = vec*weights;
    return w_vec.sum()
```

```python
def activation(val, threshold):
    if val >= threshold:
        return 1
    else:
        return 0
```

```python
#copy pasted input from above.
piv = np.array([[.4,.4], [.47,.8], [1.2,.49], [1.1, .8]]) 
labels = np.array([0, 0, 0, 1])

weights = np.array([random.uniform(-1.0,1.0), random.uniform(-1.0,1.0)])
```

```python
def fit (cycles):
    step = 2.0
    #count = 0
    for c in range(cycles):
        #intialize error to 0 for each cycle
        num_errors = 0
    
        for i in range(len(piv)):
            #count = count + 1
            #print(count)
            #pull out the weights after the 0 position
            w_sum = weighted_sum(piv[i], weights [1:]) #add bias which is first weight
            w_sum = weights[0] + w_sum
            act = activation(w_sum, threshold)
            #determine if the weights yield right answers #if so then error 0 or no error
            error = labels[i] - act
            
            #if there was an error
            if error != 0:
                #increase error count
                num_errors += 1
                #update weights to minimize error in next cycle
                weights[1] += step * error * piv[i][0]
                weights[2] += step * error * piv[i][0]
                #update bias
                weights[0] += error
        #end fit early if a cycle had no errors
        print(c, "this is the errors", num_errors)
        accuracy = (len(piv) - num_errors) / len(piv) * 100
        print(f"{c} cycles completed. Accuracy: {accuracy:.2f}%")     
    
        if num_errors == 0:
            accuracy = (len(piv) - num_errors) / len(piv) * 100
            print(f"{c} cycles completed. Accuracy: {accuracy:.2f}%")
            print(c, weights, bias)
            break
    else:
        accuracy = (len(piv) - num_errors) / len(piv) * 100
        print(f"{c} cycles completed. Accuracy: {accuracy:.2f}%")
        print("no answer found")
```
